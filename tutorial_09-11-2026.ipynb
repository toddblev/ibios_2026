````python
# Convert all letters of sequence string to uppercase, then analyze GC content

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    sequence = sequence.upper()          # all lowercase characters converted to uppercase
    gc = 0
    for base in sequence:
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100


print(gc_content("ATGCGCTGAC"))   # 60.0
print(gc_content("atgcgctgac"))   # 60.0  (was 0.0 before)
````

````python
print(gc_content(""))

# Input empty, so code attempts to divide by an integer length value of 0 (ZeroDivisionError)
````

````python
# Report ZeroDivisionError informatively; that is, it was caused by zero-length input
# (raise stops the function and reports an error, with a message that we choose)

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    sequence = sequence.upper()
    if len(sequence) == 0:               # Avoid dividing by zero
        raise ValueError("The input DNA sequence is empty.")
    gc = 0
    for base in sequence:
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100

print(gc_content(""))
````

````python
print(gc_content("42"))

# should return an ERROR, because "42" is not a DNA sequence;
# it is a string of digit characters!
# But the code returned 0.0 !  Why?
````

````python
# Check that input contains only DNA base symbols (A, C, G, or T)

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    sequence = sequence.upper()
    if len(sequence) == 0:
        raise ValueError("The sequence is empty.")
    gc = 0
    for base in sequence:
        if base not in "ACGT":           # Reject anything other than A, C, G or T characters
            raise ValueError("Invalid base: " + base)
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100

print(gc_content("42"))           # ValueError: Invalid base: 4
````

## Finding DNA sequence motifs with regular expressions
> **Regular expressions** are a mini-language for describing text patterns, provided by Python's `re` module:
> - `ATG`: exact text
> - `TAA|TAG|TGA`: the vertical bar means *or*, so TAA, TAG or TGA
> - `[AT]`: any **one** of the characters in the brackets, here A or T
> - `[AT]{3}`: `{3}` repeats the previous item 3 times, so AAA, AAT, ATA, ... TTT

````python
# re.search() looks for a pattern in a string; .start() gives the index where it begins
import re

match = re.search("GAT", "ATGCTGGATGCG")
print(match.group(), match.start())
````

````python
# re.finditer() finds every match, so we loop over them
for match in re.finditer("ATG", "ATGCATGCATG"):
    print(match.start(), match.group())
````

````python
# special symbols in patterns: | means "or", [AT] means "A or T", {3} means "3 times"
print(re.findall("TAA|TAG|TGA", "ATGTAAGGTAGCTGA"))
print(re.findall("[AT]{3}", "GGAATGGTTTCC"))
````

> **Counting positions: Python versus molecular biology**  
> The short `re` examples earlier printed Python's 0-based positions.  
> Biologists count from **1**, so Example 1 below must print `match.start() + 1`,  
> because python counts from **0** and `match.start()` is 0 for a match at the first base.

#### Example 1

````python
# Finding DNA motifs using regular expressions (RE)
import re

dna_string = "ATGCTGGATGCATTCGGTAGGTGGATATATTTGGTTGGATTATTATATAGATCGGAGACTATGA"

# Dictionary of motifs written as regular expressions
motifs = {
    "Start codon": "ATG",           # exact sequence
    "Stop codon" : "TAA|TAG|TGA"    # | = or, meaning -> TAA, TAG, TGA
}

for name, pattern in motifs.items():
    print(name)
    for match in re.finditer(pattern, dna_string):
        print("  position", match.start()+1, match.group())
````

#### Example 2

Example DNA sequence region from the *AGAMOUS* gene promoter in *Arabidopsis thaliana*

> \> *KNUCKLES* (*KNU*) gene promoter in *Arabidopsis thaliana*  
AGATACTTGAAGTTAATGGGCCATTGTGTATATTCCTTGTATATGGGCCTTAAAATAGGT
ATGTGTGGATTAAATCCAAGAAGTAATGTGAGGAAAGGCAAGAAATAATGATGAGAAGCA
AAGAATGAGAGTGATAGATAGGTAAGTAGTGAGAGAATGGAAAAAAAGAAAGAGGGTAAG
AGAAACATAGAAACCTTCCATGTTTGGCAATTTCATCTTGGAACTTGATTCACTCTCTC

> **IUPAC nucleotide codes** describe positions where more than one base is allowed,
> as in the AG binding site WWWGGHAA:
> - W = A or T (*weak*), written `[AT]`
> - H = A, C or T (not G), written `[ACT]`
> - D = A, G or T (not C), written `[AGT]` (used later, for the reverse complement)
>
> So WWWGGHAA becomes the regular expression `[AT]{3}GG[ACT]AA`.
>
> **Promoter positions** are counted relative to the transcription start site (+1),  
> so the first base of this excerpt is -1048, and the code prints `match.start() + start`.

````python
# Example case, finding AGAMOUS (AG) transcription factor binding sites in the KNUCKLES (KNU) gene promoter
# Sun et al. (2014) Science 343:1248559, Fig. 2D

# Finding DNA motifs using regular expressions (RE)
import re

# KNU promoter, -1048 to -810 relative to the transcription start site (+1)
KNU_p2 = ("AGATACTTGAAGTTAATGGGCCATTGTGTATATTCCTTGTATATGGGCCTTAAAATAGGT"
          "ATGTGTGGATTAAATCCAAGAAGTAATGTGAGGAAAGGCAAGAAATAATGATGAGAAGCA"
          "AAGAATGAGAGTGATAGATAGGTAAGTAGTGAGAGAATGGAAAAAAAGAAAGAGGGTAAG"
          "AGAAACATAGAAACCTTCCATGTTTGGCAATTTCATCTTGGAACTTGATTCACTCTCTC")

start = -1048  # coordinate of the first base

motifs = {
    "AG binding site  WWWGGHAA"  : "[AT]{3}GG[ACT]AA",
}

for name, pattern in motifs.items():
    print(name)
    for match in re.finditer(pattern, KNU_p2):
        print("   position", match.start() + start, match.group())
````

````python
# Display the promoter with the AG binding sites highlighted in color
from IPython.display import HTML, display

# re.sub() replaces every match of the pattern; \g<0> stands for the matched text itself,
# so each binding site is wrapped in an HTML <span> tag that sets a background color
highlighted = re.sub(motifs["AG binding site  WWWGGHAA"],
                     r'<span style="background-color: gold; color: black; font-weight: bold;">\g<0></span>',
                     KNU_p2)

# show the sequence in a monospace font, 60 bases per line
display(HTML('<div style="font-family: monospace; width: 60ch; word-break: break-all;">'
             + highlighted + '</div>'))
````

### Bio `Seq` objects

> **What is an object?**
> In Python, every value is an *object*: it has a type, and comes with its own methods.  
> A string is a `str` object, with methods such as `.upper()`.  
> A Biopython `Seq` object is a new type that knows about DNA, with methods such as `.reverse_complement()`.  
> We create one by calling `Seq()` on a string, and `type()` shows which kind of object we have.

````python
# create a Biopython Seq object from the DNA string we used before
from Bio.Seq import Seq

dna_string = "ATGCTGGATGCGATCTAGGAGACTATGA"
dna_seq = Seq(dna_string)

print(dna_seq)
print(type(dna_string))   # a plain Python string
print(type(dna_seq))      # a Biopython Seq object
````

````python
# a Seq object works like a string: length, indexing, slicing and counting
print(len(dna_seq))
print(dna_seq[0])
print(dna_seq[-1])
print(dna_seq[0:9])
print(dna_seq.count("G"))
````

````python
# reverse complement: our string approach from earlier, versus the Biopython method

revcomp_string = dna_string[::-1].lower().translate(str.maketrans("atcg", "tagc"))
print(revcomp_string.upper())

print(dna_seq.reverse_complement())
````

````python
# other common Seq methods
print(dna_seq.complement())    # complement, without reversing
print(dna_seq.transcribe())    # DNA -> RNA (T becomes U)
````

````python
# translate DNA into a protein sequence (* marks a stop codon)
# dna_seq is 28 nt long, so we translate the first 27 nt (9 complete codons)

print(dna_seq[0:27].translate())
print(dna_seq[0:27].translate(to_stop=True))   # stop at the first stop codon
````

````python
# display the translation with three-letter amino acid symbols
from Bio.SeqUtils import seq3

print(seq3(dna_seq[0:27].translate()))
````

````python
# translate all three reading frames on the forward strand
for frame in [0, 1, 2]:
    frame_seq = dna_seq[frame:]
    codons = len(frame_seq) // 3            # number of complete codons
    print("Frame", frame + 1, frame_seq[0:codons * 3].translate())
````

````python
# Reminder: our user-defined gc_content() function requires several lines of custom code
# (a simple version here, without the uppercase and validity checks from the functions section)

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    gc = 0
    for base in sequence:
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100

print(f"Using custom gc_content(): {gc_content(dna_seq):.1f} %")

# Using Biopython, we can calculate GC content as a fraction from 0.00 to 1.00
# requires only two lines of code

from Bio.SeqUtils import gc_fraction

print(f"Using Biopython's gc_fraction(): {gc_fraction(dna_seq) * 100:.1f} %")
````

#### Can a `Seq` object be mutated in place?

### Searching for motifs on both DNA strands
Transcription factors can bind either strand of the promoter, but our motif search above only  
scanned the forward strand. With a `Seq` object we can easily scan the reverse complement too.  

> **Why search the reverse complement?** The two DNA strands are complementary and run in opposite directions:
> ```
> + strand  5'-AAAGGCAA-3'
> - strand  3'-TTTCCGTT-5'
> ```
> Read 5'→3', the − strand above is `TTGCCTTT`, the reverse complement of the + strand.  
> So a site on the − strand appears on the + strand as its reverse complement, and searching  
> the reverse complement of the sequence finds the − strand sites.

````python
# search the TEST promoter for AGAMOUS (AG) binding sites on both strands;
# also look for restriction enzyme sites

import re
from Bio.Seq import Seq

start = -1048  # coordinate of the first base in this promoter excerpt

TEST = ("AGATACTTGAAGTTAATGGGCCTTGCCAAATATTCCTTGTATATGGGCCTTAAAATAGGT"
        "ATGTGTGGATTAAATCCAAGAAGTAATGTGAGGAAAGGCAAGAAATAATGATGAGAAGCA"
        "AAGAATGAGAGCTATAGATAGGTAAGTAGTGAGAGAATGGAAAAAAAGAAAGAGGGTAAG"
        "AGAAACATAGAAACCTTCCATGTTTGGCAATTTCATCTTGGAACTTTACCATTTCTCTC")

TEST_seq = Seq(TEST)
TEST_revcomp = TEST_seq.reverse_complement()

motifs = {
    "AG binding site  WWWGGHAA"  : "[AT]{3}GG[ACT]AA",
    "AluI restriction site"      : "AGCT",
    "HaeIII restriction site"    : "GGCC"
}

for name, pattern in motifs.items():
    print(name)
    for match in re.finditer(pattern, str(TEST_seq)):        # re needs a plain string
        print("   + strand, position", match.start() + start, match.group())
    for match in re.finditer(pattern, str(TEST_revcomp)):
        position = start + len(TEST_seq) - 1 - match.start()   # promoter coordinate of the match's first base
        print("   - strand, position", position, match.group())

# Note: for - strand matches, position is the motif's first base on that strand,
# which corresponds to the highest promoter coordinate the motif covers.

# Note: AluI (AGCT) and HaeIII (GGCC) sites are palindromic: their reverse complement
# is the same sequence (e.g. AGCT reversed is TCGA, and the complement of TCGA is AGCT).
# So each restriction site is found on BOTH strands and is listed twice: once as a
# + strand match (its lowest coordinate) and once as a - strand match (its highest).
# For example, AluI at -919 (+) and -916 (-) is one and the same 4-bp site,
# so the TEST promoter has 1 AluI site and 2 HaeIII sites, not 2 and 4.
````

````python
# Display the TEST promoter with each type of site highlighted in its own color
from IPython.display import HTML, display

# On the + strand, a - strand AG site appears as the reverse complement of WWWGGHAA,
# which is TTDCCWWW (D = A, G or T), written as a regular expression below.
# AluI (AGCT) and HaeIII (GGCC) sites are palindromes: they read the same on both strands.
highlight_colors = {
    "[AT]{3}GG[ACT]AA" : "gold",           # AG binding site, + strand
    "TT[AGT]CC[AT]{3}" : "lightskyblue",   # AG binding site, - strand
    "AGCT"             : "lightgreen",     # AluI restriction site
    "GGCC"             : "pink",           # HaeIII restriction site
}

# wrap every match of each pattern in an HTML <span> tag with its background color
highlighted = TEST
for pattern, color in highlight_colors.items():
    highlighted = re.sub(pattern,
                         rf'<span style="background-color: {color}; color: black; font-weight: bold;">\g<0></span>',
                         highlighted)

legend = ('<span style="background-color: gold; color: black;">AG site, + strand</span> '
          '<span style="background-color: lightskyblue; color: black;">AG site, - strand</span> '
          '<span style="background-color: lightgreen; color: black;">AluI</span> '
          '<span style="background-color: pink; color: black;">HaeIII</span>')

# show the legend, then the sequence in a monospace font, 60 bases per line
display(HTML('<p>' + legend + '</p>'
             '<div style="font-family: monospace; width: 60ch; word-break: break-all;">'
             + highlighted + '</div>'))
````

The next cell is **optional**: it shows how much code is needed to read FASTA records
without Biopython. Skim it, then compare it with the Biopython version that follows.

````python
# Minimal solution for reading multi-fasta using python standard library

records = []                   # list to hold header:sequence pairs
header = None                  # current header
seq = ""                       # current sequence

# open the file for reading
with open("example.fasta") as filehandle:
    for line in filehandle:
        line = line.strip()    # remove newline characters
        if not line:           # skip blank lines
            continue

        if line.startswith(">"):
            # if we just finished an upstream record, save it first
            if header:
                records.append({"header": header, "sequence": seq})
            header = line[1:]    # now, store new header (skip ">")
            seq = ""             # reset sequence
        else:
            seq += line.upper()  # add this line to ongoing sequence

# save the last record after the filehandle loop ends
if header:
    records.append({"header": header, "sequence": seq})

# print a quick summary
print(f"Loaded {len(records)} sequences from FASTA using custom code:")
for rec in records:
    print(f"{rec['header']:<18}  {len(rec['sequence']):>5} nt  start: {rec['sequence'][:30]}...")
````

````python
# More robust approach using Biopython to load multi-fasta file

from Bio import SeqIO

with open("example.fasta") as filehandle:
    records = list(SeqIO.parse(filehandle, "fasta"))   # parse() reads the records one at a time;
                                                        # list() collects them all into a list
````

#### Common methods for use with Biopython's SeqRecord and Seq objects

| Use                | Method                                 | Comments                    |
| ------------------ | -------------------------------------- | --------------------------- |
| Access id, name    | `rec.id` or `rec.name`                 | get id, name from SeqRecord |
| Access description | `rec.description`                      | get desc. from SeqRecord    |
| Access sequence    | `rec.seq`                              | get seq from SeqRecord      |
| Size & slicing     | `len()`, `[start:end]`                 | works on SeqRecord and Seq  |
| Reverse complement | `.reverse_complement()`                | works on SeqRecord and Seq  |
| Transcription      | `.transcribe()`                        | DNA → RNA conversion        |
| Translation        | `.translate()`                         | allows translation to aa seq|
| GC content         | `.count()` or `SeqUtils.gc_fraction()` | basic properties of the DNA |

### Writing FASTA files from Biopython records

````python
# Example of outputting a file from biopython records

from Bio import SeqIO

# Input and output filenames
input_file = "example.fasta"
output_file = "example_revcomp.fasta"

# Read all records, reverse complement, and write out
# (opening a file with "w" creates it, or overwrites it if it already exists)
with open(input_file) as infile, open(output_file, "w") as outfile:
    for record in SeqIO.parse(infile, "fasta"):
        rev_record = record.reverse_complement(id=True, name=True, description=True) # New record with rev_comp
        rev_record.description += " [reverse complement]"                            # add note to description
        SeqIO.write(rev_record, outfile, "fasta")                                    # Write to output FASTA

print(f"Reverse-complemented sequences written to '{output_file}'")
````

````python
# More customized file output from biopython records

from Bio import SeqIO
from Bio.SeqIO.FastaIO import FastaWriter

input_file = "example.fasta"
output_file = "example_revcomp_100bp.fasta"

rev_records = []
# Read all records, reverse complement
with open(input_file) as infile:
    for record in SeqIO.parse(infile, "fasta"):
        rev_record = record.reverse_complement(id=True, name=True, description=True)
        rev_record.description += " [reverse complement]"
        rev_records.append(rev_record)

# Open output file and use FastaWriter to choose the custom fasta wrap
with open(output_file, "w") as out_handle:
    writer = FastaWriter(out_handle, wrap=100)   # wrap lines at 100 bases
    writer.write_file(rev_records)

print(f"Wrote {len(rev_records)} reverse-complemented sequences to '{output_file}'")
````
