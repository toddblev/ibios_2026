## Defining functions: Example, GC content of a DNA sequence

```python
# Analyze a first DNA sequence to check it's GC content

seq1 = "ATGCGCTGAC"
gc = 0
for base in seq1:
    if base == "G" or base == "C":
        gc = gc + 1
print(gc / len(seq1) * 100)

# Analyze a second DNA sequence to check it's GC content

seq2 = "ATTGTTCAAT"
gc = 0
for base in seq2:
    if base == "G" or base == "C":
        gc = gc + 1
print(gc / len(seq2) * 100)

# What is the problem with this approach to analysing the two sequences?
```

```python
# Define a function to package code under a name and reuse it

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    gc = 0
    for base in sequence:
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100


print(gc_content("ATGCGCTGAC"))   # 60.0
print(gc_content("ATTGTTCAAT"))   # 20.0
print(gc_content("CCGGGCCCGG"))   # 100.0
```

```python
print(gc_content("atgcgctgac"))

# should return an 60.0, but the code instead returned 0.0 !  Why?
```

```python
# Only analyze uppercase characters, via conversion

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    sequence = sequence.upper()          # all lowercase characters converted to uppercase
    gc = 0
    for base in sequence:
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100


print(gc_content("ATGCGCTGAC"))   # 60.0
print(gc_content("atgcgctgac"))   # 60.0  (was 0.0 before)
```

```python
print(gc_content(""))

# Input empty, so code attempts to divide by an integer lenth value of 0 (ZeroDivisionError)
```

```python
# Report ZeroDivisionError informatively; that is, it was casued by zero-length input

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    sequence = sequence.upper()
    if len(sequence) == 0:               # Avoid dividing by zero
        raise ValueError("The input DNA sequence is empty.")
    gc = 0
    for base in sequence:
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100


print(gc_content("ATGCGCTGAC"))   # 60.0
print(gc_content(""))
```

```python
print(gc_content("42"))

# should return an ERROR, because input was an integer! But the code returned 0.0 !  Why?
```

```python
# Check that input contains only DNA base symbols (A, C, G, or T)

def gc_content(sequence):
    """Return the percentage of G and C bases in a DNA sequence."""
    sequence = sequence.upper()
    if len(sequence) == 0:
        raise ValueError("The sequence is empty.")
    gc = 0
    for base in sequence:
        if base not in "ACGT":           # Reject anything other than A, C, G or T characters
            raise ValueError("Invalid base: " + base)
        if base == "G" or base == "C":
            gc = gc + 1
    return gc / len(sequence) * 100


print(gc_content("ATGCGCTGAC"))   # 60.0
print(gc_content("42"))           # ValueError: Invalid base: 4
```
